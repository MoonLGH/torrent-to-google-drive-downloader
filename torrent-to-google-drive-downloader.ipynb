## Mount Google Drive

## Helper: Generate folder download otomatis berdasarkan timestamp

```python
from datetime import datetime
from pathlib import Path

BASE_DIR = '/content/drive/My Drive/Torrent'

def make_download_dir():
    now = datetime.now().strftime('%Y-%m-%d_%H-%M-%S')
    path = Path(BASE_DIR) / now
    path.mkdir(parents=True, exist_ok=True)
    print('Folder download:', path)
    return str(path)
```

## Download dari File .torrent

```python
from google.colab import files

download_dir = make_download_dir()

print('Upload file .torrent kamu...')
uploaded = files.upload()

for name in uploaded.keys():
    print(f'Mulai download dari .torrent: {name}')
    !aria2c --seed-time=0 --dir="{download_dir}" "{name}"

print('Selesai menjalankan semua .torrent.')
```

## Download dari Magnet Link

```python
download_dir = make_download_dir()

while True:
    magnet_link = input('Masukkan Magnet Link atau ketik "exit": ').strip()
    if magnet_link.lower() == 'exit':
        print('Keluar dari input magnet.')
        break
    if not magnet_link:
        continue

    print('Mulai download magnet...')
    !aria2c --seed-time=0 --dir="{download_dir}" "{magnet_link}"
```
